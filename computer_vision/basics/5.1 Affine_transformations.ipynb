# 2-D transformations (Affine)

Transformation could be represented as a matrix-vector multiplication.

Convention: usually matrix (on the left) is multiplied by a vector-column (on the right).

## Geometric interpretation

Columns of the matrix represent the new basis vectors after transformation.
Example:
- $\vec{i}$ = $\begin{bmatrix} 1 \\ 0 \end{bmatrix}$
- $\vec{j}$ = $\begin{bmatrix} 0 \\ 1 \end{bmatrix}$
- A = $\begin{bmatrix} a & b \\
                        c & d \end{bmatrix}$
- $A \cdot i$ = $\begin{bmatrix} a \cdot 1 + b \cdot 0 \\ c \cdot 1 + d \cdot 0 \end{bmatrix}$ = $\begin{bmatrix} a \\ c \end{bmatrix}$ – the new position of the vector $\vec{i}$ after the transformation.
- $A \cdot j$ = $\begin{bmatrix} a \cdot 0 + b \cdot 1 \\ c \cdot 0 + d \cdot 1 \end{bmatrix}$ = $\begin{bmatrix} b \\ d \end{bmatrix}$ – the new position of the vector $\vec{j}$ after the transformation.


Any matrix simply tells the space: "Move the X arrow here and the Y arrow there, and uniformly stretch the rest of the space along with them.

## Orthonormal matrix

If a matrix $A$ is orthonormal, it transforms the space in the least aggressive way possible: no skew or scale applied, just rotation or reflection.

Orthonormal matrix is defined as: $A^TA=I$

Consequences:
1. Determinant is always +1 or -1:

   $|A^TA|=|A|\cdot|A|=|A|^2$

   $A^TA=I; |I|=1 ⇒ |A^TA|=1 ⇒ |A|^2=1 ⇒ |A|=±1$
2. Space deformation:
    1. $|A|=1$ – clean rotation. The space simply rotates around the beginning of coordinates by degree $\theta$.
    2. $|A|=-1$ – reflection. The space is reflected relative to the reflection axis.
3. Invariants:
    - Lengths are preserved
    - Distances are preserved
    - Angles are preserved


Note: if the transformation matrix is orthogonal (not orthonormal), it also scales the space, but no skew is applied.

## Determinant

**Determinant** – a value calculated for a square matrix that shows exactly how the linear transformation defined by that matrix changes the scale of the space (area, volume, or orientation).
In other words, it is the "scaling factor" of space.


Imagine a standard coordinate grid (where a unit square or cube has an area or volume of 1). When we multiply vectors by a matrix, we transform (shear, rotate, stretch) this space.
- 2D space (2x2 matrix): The determinant equals the area of parallelogram into which the initial unit square is transformed.
- 3D space (3x3 matrix): The determinant equals the volume of the parallelepiped into which the initial unit cube is transformed.
- Higher dimensions (nD): It is the hypervolume of the corresponding figure (parallelepiped/parallelotope).

Meaning of the value of the determinant det(A):
- det(A) = 2: The transformation doubles any area or volume.
- det(A) = 1: Space may rotate or shear, but the total volume or area remains unchanged (e.g., a pure rotation using a rotation matrix).
- det(A) = 0 (Singular matrix): This means the transformation "flattens" space into a lower dimension. For example, an entire 2D plane is compressed into a single line or point. Since a line has an area of 0, the determinant is also 0.
    - Why is this important for systems of equations? If space has been flattened, information is lost, and the process cannot be reversed (one cannot unfold the line back into a plane). Consequently, no inverse matrix exists for such matrices, and systems of equations either have no solutions or infinitely many solutions.
- Negative determinant det(A) < 0: The minus sign indicates that space has not merely stretched but has changed its orientation (a reflection has occurred).
    - In 2D: if you were moving along a circle in a clockwise direction, a transformation with a negative determinant would result in counter-clockwise movement (space has effectively been "turned inside out").
