# 2-D transformations (Affine)

Transformation could be represented as a matrix-vector multiplication.

Convention: usually matrix (on the left) is multiplied by a vector-column (on the right).

## Geometric interpretation

Columns of the matrix represent the new basis vectors after transformation.
Example:
- $\vec{i}$ = $\begin{bmatrix} 1 \\ 0 \end{bmatrix}$
- $\vec{j}$ = $\begin{bmatrix} 0 \\ 1 \end{bmatrix}$
- A = $\begin{bmatrix} a & b \\
                        c & d \end{bmatrix}$
- $A \cdot i$ = $\begin{bmatrix} a \cdot 1 + b \cdot 0 \\ c \cdot 1 + d \cdot 0 \end{bmatrix}$ = $\begin{bmatrix} a \\ c \end{bmatrix}$ – the new position of the vector $\vec{i}$ after the transformation.
- $A \cdot j$ = $\begin{bmatrix} a \cdot 0 + b \cdot 1 \\ c \cdot 0 + d \cdot 1 \end{bmatrix}$ = $\begin{bmatrix} b \\ d \end{bmatrix}$ – the new position of the vector $\vec{j}$ after the transformation.


Any matrix simply tells the space: "Move the X arrow here and the Y arrow there, and uniformly stretch the rest of the space along with them.

## Orthonormal matrix

If a matrix $A$ is orthonormal, it transforms the space in the least aggressive way possible: no skew or scale applied, just rotation or reflection.

Orthonormal matrix is defined as: $A^TA=I$

Consequences:
1. Determinant is always +1 or -1:

   $|A^TA|=|A|\cdot|A|=|A|^2$

   $A^TA=I; |I|=1 ⇒ |A^TA|=1 ⇒ |A|^2=1 ⇒ |A|=±1$
2. Space deformation:
    1. $|A|=1$ – clean rotation. The space simply rotates around the beginning of coordinates by degree $\theta$.
    2. $|A|=-1$ – reflection. The space is reflected relative to the reflection axis.
3. Invariants:
    - Lengths are preserved
    - Distances are preserved
    - Angles are preserved


Note: if the transformation matrix is orthogonal (not orthonormal), it also scales the space, but no skew is applied.

# Transition from one basis to another

1. Transition matrix $P$ (new → old basis) consists of the new basis vectors placed in columns in terms of the coordinates of the old basis.
2. Find the inverse matrix $P^{-1}$ (old → new basis).
3. Transition equation: $X_{new} = P^{-1} \cdot X_{old}$

# Determinant

**Determinant** – a value calculated for a square matrix that shows exactly how the linear transformation defined by that matrix changes the scale of the space (area, volume, or orientation).
In other words, it is the "scaling factor" of space.


Imagine a standard coordinate grid (where a unit square or cube has an area or volume of 1). When we multiply vectors by a matrix, we transform (shear, rotate, stretch) this space.
- 2D space (2x2 matrix): The determinant equals the area of parallelogram into which the initial unit square is transformed.
- 3D space (3x3 matrix): The determinant equals the volume of the parallelepiped into which the initial unit cube is transformed.
- Higher dimensions (nD): It is the hypervolume of the corresponding figure (parallelepiped/parallelotope).


<img src="0_images/determinant_2d.png" width="350">


Meaning of the value of the determinant det(A):
- det(A) = 2: The transformation doubles any area or volume.
- det(A) = 1: Space may rotate or shear, but the total volume or area remains unchanged (e.g., a pure rotation using a rotation matrix).
- det(A) = 0 (Singular matrix): This means the transformation "flattens" space into a lower dimension. For example, an entire 2D plane is compressed into a single line or point. Since a line has an area of 0, the determinant is also 0.
    - Why is this important for systems of equations? If space has been flattened, information is lost, and the process cannot be reversed (one cannot unfold the line back into a plane). Consequently, no inverse matrix exists for such matrices, and systems of equations either have no solutions or infinitely many solutions.
- Negative determinant det(A) < 0: The minus sign indicates that space has not merely stretched but has changed its orientation (a reflection has occurred).
    - In 2D: if you were moving along a circle in a clockwise direction, a transformation with a negative determinant would result in counter-clockwise movement (space has effectively been "turned inside out").

# Eigenvectors & Eigenvalues

## Definition

Any matrix defines a space transformation.
Each vector of the space multiplied by the matrix changes its direction and magnitude.

**Eigenvectors ($\vec{v}$)** are special vectors that preserve their direction but change magnitude (or turn in the opposite direction).

**Eigenvalues ($\lambda$)** - scaling coefficients of eigenvectors.

$\qquad A \cdot \vec{v} = \lambda \cdot \vec{v}$

$\qquad \vec{v}\neq\vec{0}$

## Calculations:
1. $A \vec{v} = \lambda I \vec{v}$ - transform $\lambda$ from number to diagonal matrix
2. $A \vec{v} - \lambda I \vec{v} = \vec{0}$
3. $(A - \lambda I)\vec{v} = \vec{0}$
4. $\vec{v}\neq\vec{0} \implies det(A - \lambda I) = 0$

    If a matrix is multiplied by a non-zero vector and the result is zero,
    this means that the matrix is singular (non-invertible). This means that one or more dimensions are lost.
    An indication of matrix singularity is a determinant's quality to zero.

5. $\lambda_1, ..., \lambda_n$ - eigenvalues, solutions of (4)
6. To find eigenvectors $\vec{v_i}$, for each $\lambda_i, i=\overline{1,n}$ solve the equation (3):

    $(A - \lambda_i I)\vec{v_i} = \vec{0}$

## Examples
1. **Anisotropic Scaling:**

    $A=\begin{bmatrix} 3 & 0 \\ 0 & 0.5 \end{bmatrix}$

    Eigenvectors:

    $\vec{v}_1=\begin{bmatrix} 1 \\ 0 \end{bmatrix}$; $\vec{v}_2=\begin{bmatrix} 0 \\ 1 \end{bmatrix}$

    Eigenvalues:

    $\lambda_1=3; \lambda_2=0.5$

    Any vector will change the direction, but vectors collinear to $\vec{v}_1$ and $\vec{v}_2$ will only change magnitude by a factor of $\lambda_1$ and $\lambda_2$, respectively.

2. **Diffuse stretching along the diagonals:**

    $A=\begin{bmatrix} 1.25 & 0.75 \\ 0.75 & 1.25 \end{bmatrix}$

    Eigenvectors (with angle $45^\circ$):

    $\vec{v}_1=\begin{bmatrix} 1 \\ 1 \end{bmatrix}$; $\vec{v}_2=\begin{bmatrix} 1 \\ -1 \end{bmatrix}$

    Eigenvalues:

    $\lambda_1=2; \lambda_2=0.5$

3. **Rotation** - all vectors change direction. Eigenvalues are complex.

# Correlation

Pearson correlation coefficient – measure of linear relation. It is used the most often.

Calculation algorithm:
1. 2 feature vectors of size n: X and Y
2. Calculate mean and std for X and Y: $\overline{X}, \sigma_X, \overline{Y}, \sigma_Y$
3. Calculate covariation:

    $cov(X,Y) = \frac{\sum_{i=1}^{n}(X_i - \overline{X})(Y_i - \overline{Y})}{n-1}$

4. Calculate Pearson correlation coefficient:

    $r = \frac{cov(X,Y)}{\sigma_X \sigma_Y}$

5. $r\in [-1,1]$

Basically, it is multiplication of NORMALIZED vectors, normed afterward.

Intuition: the maximum response will be if values are both positive or both negative.

# SVD – Singular Value Decomposition

We have matrix A with size $m x n$.

We cannot compute eigenvectors for rectangular matrices, that is why we use SVD.

The matrix A is decomposed into 3 simpler geometric operations:
1. Rotation
2. Scaling (+ dimension change)
3. Rotation

Formula:

$\qquad A=U \cdot \Sigma \cdot V^T$

Where:
- $V (n \times n)$ - right singular vectors:
    1. Create matrix $A^T A$ ($n \times n$ dimensions), has meaning of covariance matrix.
    2. Find eigenvalues $\lambda_i$ and eigenvectors ($v_i$) of $A^T A$
    3. Normalize eigenvectors
  Calculated eigenvectors form columns of $V$.
- $\Sigma (m \times n)$ - singular values
    1. $\sigma_i=\sqrt{\lambda_i}$ where $\lambda_i$ is eigenvalues of $A^T A$ from matrix $V$ calculation.
    2. Singular values are sorted from largest to smallest. They are placed on the main diagonal of $\Sigma$

    (main diagonal elements: $\Sigma_ii, i=\overline{1,n}$, does not end up in the bottom-right corner for rectangular matrices).
- $U (m \times m)$ - left singular values
    1. Approach A: calculate eigenvectors of $AA^T$
    2. Approach B: $u_i = \frac{1}{\sigma_i}Av_i$, where $v_i$ and $\sigma_i$ are right singular vectors and singular values.

# PCA – Principal Components Analysis

PCA uses eigenvectors as target vectors for data points projection because it allows to reduce a noise and unimportant information:
1. Eigenvectors show directions of maximum data dispersion.
2. Eigenvalues show the value of dispersion for each eigenvector → the volume of information.
3. Eigenvectors of symmetric covariation matrix are orthogonal. The correlation between new axes is zero.
Each axis contains unique information, not duplicated in other axes.


# Matrix properties

- Determinant – see above.
- Eigenvectors, Eigenvalues, Spectral vectors, Spectral values – see above.
- Spectral radius – maximum absolute eigenvalue of a matrix. Represent a scaling dynamic of vectors multiplied by $W$.

    $\rho(W)=max_i{|\lambda_i|}$

    - $\rho(W)>1$ – multiplying vectors sequentially by \(W\) causes them to grow exponentially (gradient explosion).
    - $\rho(W)<1$ – multiplying vectors sequentially by \(W\) causes them to decay exponentially (gradient vanishing).

    - **NOTE:** for non-symmetric weight matrices, the maximum and minimum bounds of expansion or contraction are defined by the maximum and minimum singular values $(\sigma_{\max}, \sigma_{\min})$. Orthogonal initialization of weights in NN aims for all singular values to equal 1 (\sigma_i = 1), acting as a rigid rotation that prevents scaling artifacts altogether. In other words:

        $\qquad\sigma _{\min }\|{}x\|{}\le \|{}Wx\|{}\le \sigma _{\max }\|{}x\|{}$

- Condition Number - measures how sensitive the output $x$ to errors or floating-point noise in $A$ or $b$ for SLAR $Ax = b$:

    $k(A)=\|A\| \cdot \|A^{-1}\|=\frac{\sigma_{max}}{\sigma_{min}}$

    - $k(A)~1$ - matrix is well-conditioned, a small error in the input creates a proportionally small error in the output.
    - $k(A) >> 1$ - matrix is ill-conditioned, a small error in the input can cause a large error in the output.

    **Note**: while estimating a homography or fundamental matrix, we need to solve an overdetermined SLAR.
    Because components have dimensionality of pixels (from range $[0,1920]$ or larger), the matrix $A$ has astronomical condition number.
    To mitigate this issue Hartley's Normalization is applied. Before building the matrix, keypoints are translated so their centroid is at the origin $(0,0)$ and scaled so their average distance to the origin is $\sqrt{2}$.
- Optical Flow & Feature Tracking (The Lucas-Kanade Matrix).
    To track a pixel's motion, the Lucas-Kanade algorithm solves a small \(2 \times 2\) linear system using a spatial gradient matrix:

    $H=\left[\begin{matrix}\sum I_{x}^{2}&\sum I_{x}I_{y}\\ \sum I_{x}I_{y}&\sum I_{y}^{2}\end{matrix}\right]$

    The eigenvalues $(\lambda_1, \lambda_2)$ of this matrix describe the texture under the tracking window:
    - **Both eigenvalues are large** – the matrix is **stable** (low condition number), indicating a distinct corner that can be tracked reliably.
    - **One eigenvalue is near zero** – the matrix is **ill-conditioned**, meaning it's an edge where motion along the edge cannot be determined (the Aperture Problem).
    - **Both are near zero** – it's a flat, textureless region where tracking is entirely unstable.
- Essential Matrix $E$ Singularity Enforcement – an Essential Matrix  models 3D camera geometry between two views.
It has a strict structural definition: its singular values must be $(\sigma, \sigma, 0)$ — two identical values and one zero.
    - Due to noise in pixel detection, an estimated matrix $\hat{E}$ will never perfectly yield these singular values.
    - Stability is enforced by performing an SVD on the raw estimate $(\hat{E} = U \Sigma V^T)$, manually resetting the diagonal matrix to $\Sigma = \text{diag}(1, 1, 0)$, and re-multiplying. Without this explicit projection step, the 3D reconstruction decomposes into non-rigid, distorted geometry.
- Covariance Matrix Singularity in 3D Gaussian Splatting / SLAM - In modern vision tasks like 3D Gaussian Splatting or Kalman filtering for SLAM, 3D shapes and uncertainties are tracked via spatial Covariance Matrices $(\Sigma)$.
    - When a Gaussian becomes exceptionally flat or an optimization converges perfectly in one direction, the smallest eigenvalue drops to absolute zero, causing numerical explosion during the required matrix inversion (\(\Sigma ^{-1}\)).
    - Stability Regularization: Algorithms enforce stability by adding a tiny diagonal isotropic perturbation $(\Sigma + \epsilon I)$, physically preventing the spatial volume from collapsing into a zero-thickness singularity.
