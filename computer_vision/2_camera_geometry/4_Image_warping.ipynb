# Image warping

If we have several images taken by the same camera with the same location but with different rotation angles, we can project one image on another.

Note: such an approach is simpler than 3D reprojection.

Intuition: cast ray through each pixel of img1 → draw the pixel where that intersects img2.

Note: this model only holds for angular views up to 180$^\circ$. Beyond that need to use sequence that "blends the rays" or map onto a different surface, say, cylinder.

The solution is to use homography transformation:
1. Select 4 pairs of corresponding points on each image.
2. Find homography matrix:
    1. For non-homogeneous coordinates: as h for $\min_h \lVert Ah-b \rVert^2$ using least-squares
    2. For homogeneous coordinates: as SVD decomposition (like in camera calibration theory)
3. Map each pixel of the 2nd image to the 1st image surface using the homography matrix. There are 2 ways of mapping:
    1. For each pixel of the source image compute the location of the destination pixel and put the color value there. If it falls between pixels – distribute color value among neighbors. This is not the recommended way.
    2. For each pixel of the destination image compute the location of the source pixel and take the color value from that pixel. If it falls between pixels – use linear interpolation (or near neighbor) to estimate the color value at the target location. This is the recommended way.
