# Image to image projection

There are 5 types of projection:

<img src="3_images/transformation_types.png" width="1000">

In homogeneous coordinates all of them could be represented as matrix multiplication and will have the next number of DoF (hence the number of points we need to estimate parameters):
- Translation - 2 DoF
- Euclidean (rigid body) - 3 DoF (translation + rotation)
- Similarity - 4 DoF (euclidean + scale)
- Affine - 6 DoF (euclidean + skew) (parallel lines are still parallel)
- Homography (general projective transform) - 8 DoF (Affine + perspective)
