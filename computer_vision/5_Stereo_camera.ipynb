# Stereo camera advantages

For Visual Odometry and SLAM:
- Allows directly estimating the depth of observed points (epipolar geometry)
- Allows recovering camera motion in a real-world metric scale
- Better handling of pure rotation.

Advantages for VO:
- Robust motion estimation (based on PnP + RANSAC instead of Essential matrix estimation)


For SLAM:
- Quicker map initialization.
    Monocular SLAM typically needs $2$ or more frames with sufficient parallax to initialize its map.
    With stereo this estimate is much quicker ($1$ frame or more).
- Additional geometric constraints during bundle adjustment.


# Stereo camera disadvantages

- Depth estimation becomes less accurate as distance increases.
    Approximately:
    $$
    \sigma_Z \approx \frac{Z^2}{fB}\sigma_d
    $$
    Depth uncertainty grows quadratically with distance.

    At 2 m, depth uncertainty is around 2.4 cm. At 20 m, it is approximately 2.38 m under the same assumptions.
- Increased hardware and processing requirements.
- Complex synchronization.
- More complex calibration.
- Stereo provides metric scale, but it does not guarantee lower drift than monocular SLAM in every environment
